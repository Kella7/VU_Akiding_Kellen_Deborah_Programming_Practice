````python
# classes and objects
class Patient:

    def __init__(self, name, age, diagnosis):

        self.name = name
        self.age = age
        self.diagnosis = diagnosis

    def display_details(self):

        print("Patient:", self.name)
        print("Age:", self.age)
        print("Diagnosis:", self.diagnosis)


patient1 = Patient(
    "Sarah",
    30,
    "Malaria"
)

patient1.display_details()
````

````python
#inheritance
class Employee:

    def __init__(self, name, salary):

        self.name = name
        self.salary = salary

    def display_employee(self):

        print("Employee:", self.name)
        print("Salary:", self.salary)


class Doctor(Employee):

    def __init__(self, name, salary, speciality):

        # Reuse the parent constructor.
        super().__init__(name, salary)

        self.speciality = speciality

    def display_doctor(self):

        self.display_employee()

        print("Speciality:", self.speciality)


doctor1 = Doctor(
    "Dr Sarah",
    5000000,
    "Paediatrics"
)

doctor1.display_doctor()
````

````python
#encapsulation
class BankAccount:

    def __init__(self, owner, balance):

        self.owner = owner

        # Double underscore indicates internal/private-style data.
        self.__balance = balance

    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount

    def get_balance(self):

        return self.__balance


account = BankAccount(
    "Deborah",
    500000
)

account.deposit(200000)

print(account.get_balance())
````

````python
#supervised classification example
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

ml_data = {
    "Age": [
        25, 30, 35, 40,
        45, 50, 55, 60,
        28, 48, 58, 33
    ],

    "BMI": [
        20, 22, 25, 28,
        30, 35, 32, 37,
        21, 31, 36, 24
    ],

    "Glucose": [
        80, 85, 95, 110,
        130, 160, 150, 180,
        82, 140, 170, 90
    ],

    # 0 = No diabetes
    # 1 = Diabetes
    "Diabetes": [
        0, 0, 0, 0,
        1, 1, 1, 1,
        0, 1, 1, 0
    ]
}

ml_df = pd.DataFrame(ml_data)

X = ml_df[
    [
        "Age",
        "BMI",
        "Glucose"
    ]
]

y = ml_df["Diabetes"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

model = LogisticRegression(
    max_iter=1000
)

model.fit(
    X_train,
    y_train
)

predictions = model.predict(
    X_test
)

print(
    "Accuracy:",
    accuracy_score(
        y_test,
        predictions
    )
)

print(
    "\nConfusion Matrix:"
)

print(
    confusion_matrix(
        y_test,
        predictions
    )
)

print(
    "\nClassification Report:"
)

print(
    classification_report(
        y_test,
        predictions,
        zero_division=0
    )
)
````

````python
#regression example
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

crop_data = {
    "Rainfall": [
        700,
        800,
        850,
        900,
        1000,
        1100,
        1200,
        1300
    ],

    "Fertilizer": [
        20,
        25,
        30,
        35,
        40,
        45,
        50,
        55
    ],

    "Yield": [
        1.5,
        1.8,
        2.0,
        2.4,
        2.8,
        3.1,
        3.5,
        3.9
    ]
}

crop_df = pd.DataFrame(crop_data)

X = crop_df[
    [
        "Rainfall",
        "Fertilizer"
    ]
]

y = crop_df["Yield"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

regression_model = LinearRegression()

regression_model.fit(
    X_train,
    y_train
)

predictions = regression_model.predict(
    X_test
)

mae = mean_absolute_error(
    y_test,
    predictions
)

mse = mean_squared_error(
    y_test,
    predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    predictions
)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
````

````python
#Tensorflow neural network

import numpy as np
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = np.array(
    [
        [25, 20, 80],
        [30, 22, 85],
        [35, 25, 95],
        [40, 28, 110],
        [45, 30, 130],
        [50, 35, 160],
        [55, 32, 150],
        [60, 37, 180],
        [28, 21, 82],
        [48, 31, 140],
        [58, 36, 170],
        [33, 24, 90],
        [38, 27, 105],
        [52, 34, 155],
        [27, 22, 88],
        [62, 38, 190]
    ],
    dtype=float
)

y = np.array(
    [
        0, 0, 0, 0,
        1, 1, 1, 1,
        0, 1, 1, 0,
        0, 1, 0, 1
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train = scaler.fit_transform(
    X_train
)

X_test = scaler.transform(
    X_test
)

model = tf.keras.Sequential(
    [

        tf.keras.Input(
            shape=(3,)
        ),

        tf.keras.layers.Dense(
            16,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            8,
            activation="relu"
        ),

        tf.keras.layers.Dense(
            1,
            activation="sigmoid"
        )
    ]
)

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    X_train,
    y_train,
    epochs=20,
    validation_split=0.2,
    verbose=0
)

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print("Test loss:", test_loss)
print("Test accuracy:", test_accuracy)
````

````python
#making predictions with the Neural Network
probabilities = model.predict(
    X_test,
    verbose=0
)

predicted_classes = (
    probabilities >= 0.5
).astype(int)

print("Probabilities:")
print(probabilities)

print("\nPredicted classes:")
print(predicted_classes)
````

````python
#assignment
def get_grade(mark):

    if mark >= 80:
        return "A"

    elif mark >= 70:
        return "B"

    elif mark >= 60:
        return "C"

    elif mark >= 50:
        return "D"

    else:
        return "F"


def add_student(students, name, mark):

    # Validate mark.
    if mark < 0 or mark > 100:
        raise ValueError(
            "Mark must be between 0 and 100."
        )

    grade = get_grade(mark)

    student = {
        "name": name,
        "mark": mark,
        "grade": grade
    }

    students.append(student)


def display_students(students):

    if len(students) == 0:

        print(
            "No student records available."
        )

        return

    for student in students:

        print(
            f"Name: {student['name']}"
        )

        print(
            f"Mark: {student['mark']}"
        )

        print(
            f"Grade: {student['grade']}"
        )

        print("-" * 30)


students = []

add_student(
    students,
    "Deborah",
    88
)

add_student(
    students,
    "Mary",
    94
)

add_student(
    students,
    "John",
    73
)

display_students(students)
````
